# **转录组数据提交指南**

本文档说明如何准备 metadata，并将转录组数据提交至 NCBI（National Center for Biotechnology Information）、SCEA（Single Cell Expression Atlas）和 HCA Data Portal（Human Cell Atlas Data Portal）。内容适用于单细胞、bulk RNA-seq 和空间转录组数据，并遵循 FAIR（Findable, Accessible, Interoperable, Reusable）原则。


## **转录组数据中的 metadata**

Metadata 是“描述数据的数据”。换句话说，它是伴随原始数据和处理后数据一起提供的描述性信息，使数据能够被理解、放入正确背景中解释，并在后续被重复使用。Metadata 可以是包含个体或患者信息的表格，也可以是记录样本处理和文件处理细节的补充文件等。

在转录组研究中，无论是 bulk RNA-seq、单细胞还是空间转录组，metadata 都应同时覆盖生物学信息和技术信息，包括：

* 生物学信息：物种、组织、细胞类型、实验条件、处理方式、采集时间、临床或环境特征。

* 技术信息：测序平台、文库构建方案、捕获参数（例如 10x Genomics、Smart-seq2、Visium）、RNA 质量、使用的软件及版本。

* 实验背景：研究设计、比较组、重复、批次因素、对照。

* 处理后数据：count matrices、归一化文件、空间坐标等输出文件，以及使用过的脚本和流程。

结构良好的 metadata 对数据共享至关重要。它能确保其他研究者正确解释数据、验证结果，并将其整合进比较分析或参考数据库。


## **FAIR Metadata**

为了让 metadata 真正发挥作用，它需要具备一系列特征，以保证实用性和质量。在这方面，FAIR 原则提供了一组最佳实践，可帮助我们更有效地组织 metadata。

* Findable（可发现）：使用持久标识符。例如文章通常使用 DOI；数据和 metadata 可使用 BioProject、BioSample、SRA、GEO 等标识符，并配合标准化关键词以便检索。

* Accessible（可访问）：数据应存放在公共仓库中，并采用开放且机器可读的格式，保证广泛且透明的访问。

* Interoperable（可互操作）：采用公认的词汇表和本体，例如 Cell Ontology、Uberon、Disease Ontology，使不同数据库之间能够整合，并保证数据含义一致。

* Reusable（可复用）：metadata 应完整，并提供清晰的实验方案和充分文档，以支持重新分析和比较；同时也要确保个体编码和匿名化信息得到保留。

虽然这些原则并非在所有场景中强制要求，但强烈建议遵循它们，以提高数据的可见性、可访问性和复用价值。


# 提交文件格式

在理解 NCBI 提交流程之前，有必要了解最常见的文件格式。每种扩展名都承载特定类型的信息，并在 RNA-seq 分析中有不同用途，包括 bulk、单细胞或空间转录组。


### `.csv` 文件（Comma-Separated Values）

`.csv` 是字段用逗号分隔的文件，广泛被电子表格软件和分析软件接受，常用于 count 文件。

注意：如果文本本身包含逗号，可能造成字段冲突。

* Excel：File > Save As > csv

* Google Sheets：File > Download > csv

### `.tsv` 文件（Tab-Separated Values）

`.tsv` 是每列用制表符（TAB）分隔的文本文件，每一行代表一条记录，例如一个样本或一个基因。

它非常常用于 metadata 和 count matrices，因为它可以避免文本中逗号带来的冲突。

|.csv|.tsv|
|-------------|--------------|
|字段之间使用逗号|字段之间使用制表符（TAB）|
|如果文本包含逗号，可能造成冲突|对文本型 metadata 更安全|
|扩展名：.csv|扩展名：.tsv|

保存方式：

* Excel：File > Save As > Text (tab delimited) (.txt) → 如有需要，将文件重命名为 `.tsv`

* Google Sheets：File > Download > Tab delimited values (.tsv)

### `.txt` 文件（Plain Text）

纯文本文件，通常用于样本 count 或基础 metadata。它可以组织为表格形式（用 tab 或空格分隔），也可以是列表。

### `.rds` 文件（R Data Serialization）

`.rds` 是 R 专用的二进制格式。它可以保存复杂对象，例如归一化矩阵、Seurat 对象或 SingleCellExperiment 对象，同时保持其结构和 metadata。该格式非常适合在 R 分析中直接复用。

### `.R` 文件（R Script）

用 R 编写的代码文件，即 R 脚本。常用于共享分析流程，例如归一化、DESeq2、Seurat 分析等。

### `.sh` 文件（Shell Script）

`.sh` 是 shell 脚本，也就是终端脚本（bash/shell）。它常用于自动化处理步骤，例如比对、格式转换，或向服务器提交作业。

### `.ipynb` 文件（Jupyter Notebook）

交互式文件，可同时包含代码（Python、R 等）、结果和文档。它对可重复性非常有用，因为可以逐步展示分析过程。GEO 也接受 `.ipynb` 作为共享完整流程的一种方式。


## **NCBI 提交流程**


```python
[ BioProject ]
     ↓
[ BioSample ] → [ SRA (raw files: FASTQ) ]
     ↓
[ GEO (Processed data: matrices, metadata, scripts) ]
```

NCBI 使用层级结构连接不同层次的信息：

**BioProject：** 最高层级，代表整个研究。每个 project 会获得一个唯一标识符，例如 `PRJNA123456`，并把所有相关样本和实验组织在一起。

**BioSample：** 描述 project 内的每个独立生物样本，包括物种、组织、细胞类型、实验条件、处理方式和采集地点等信息。每个 BioSample 都会获得自己的编号，例如 `SAMN45678901`。

**SRA（Sequence Read Archive）：** 存储原始测序数据，例如 FASTQ 或 BAM 文件。每次提交会生成一个 `SRRxxxxxxx` 标识符。

**GEO（Gene Expression Omnibus）：** 存储处理后的数据，例如 count matrices、归一化文件和表达 metadata。每次提交会获得一个 `GSExxxxxx` 标识符。


## 创建 BioProject


BioProject 用于归集一个研究中的所有样本和数据。*每个研究只创建一个 BioProject。*

执行该步骤需要先创建 NCBI 账号。


需要填写 title、description、organism 和 data type。

示例：

project_title: Single-cell transcriptomic atlas of PBMCs during CHIKV infection

data_type: Transcriptome (bulk + single-cell + spatial)

* 通过网页创建：https://submit.ncbi.nlm.nih.gov/subs/bioproject
* 提交后会生成一个 PRJNAxxxxx 编号，例如 `PRJNA123456`。


## 创建 BioSample


每个生物样本都会获得一个唯一 ID。

对于转录组样本，应描述以下字段：

`sample_name`、`organism`、`tissue`、`cell_type`、`disease`、`treatment`、`time_point`、`geo_loc_name`

可补充字段包括：`sequencing_protocol`、`dissociation_method`、`library_prep`、`cell_capture_platform`（例如 10x Genomics Chromium、Smart-seq2）。

## **简化提交流程图**


```python
PRJNA123456   (BioProject)
   ├── SAMN45678901   (BioSample)
   │      ├── SRRxxxxxxx   (SRA - raw data)
   │      └── GSMxxxxxxx   (GEO - processed data)
   ├── SAMN45678902   (BioSample)
   │      ├── SRRyyyyyyy   (SRA - raw data)
   │      └── GSMyyyyyyy   (GEO - processed data)
```

1. 创建 BioProject

* 访问：https://submit.ncbi.nlm.nih.gov/subs/bioproject

* 按上文说明填写研究信息，例如 title、organism、type。

* 提交后会生成类似 `PRJNA123456` 的编号。

2. 提交 BioSamples

访问：https://submit.ncbi.nlm.nih.gov/subs/biosample

填写每一行时，无论通过网页表单还是 `.tsv`，都应包含以下字段：


```python
bioproject_accession
PRJNA123456
```

每个样本会获得类似 `SAMN45678901` 的编号。

每个 BioSample 必须有唯一名称，例如 CHIKV_01；同一个名称也会用于 SRA 和 GEO metadata。

3. 提交至 SRA（原始数据）

访问：https://submit.ncbi.nlm.nih.gov/subs/sra

上传 `.fastq.gz` 文件。

在 `.tsv` 文件或表单中包含：

| sample_name   |  biosample_accession |
|---------------|----------------------|

4. 提交至 GEO

进入 BioProject → 查看 BioSamples → 访问 SRA 中的数据 → 查看 GEO 中的处理后文件。整体体验就像一个相互连接的完整研究。

提交处理后文件时：

例如提交 gene counts。

* metadata 表格。

## **提交至 Single Cell Expression Atlas（SCEA）的流程**

Single Cell Expression Atlas 是 EMBL-EBI 的公共数据库，汇集单细胞 RNA-seq 和空间转录组数据。这些数据会通过标准化流程重新处理，并使用本体进行增强。提交遵循 MAGE-TAB 标准，即 IDF 和 SDRF 文件，并在整合进 Atlas 前经过人工整理。

这是[官方技术指南](https://github.com/ebi-gene-expression-group/expression-atlas-curation-guide/blob/master/pages/single_cell_technical_guide.md?form=MG0AV3&utm_source=copilot.com)，另有[补充说明](https://www.ebi.ac.uk/gxa/sc/help.html?form=MG0AV3&utm_source=copilot.com)。下面是对流程的简化说明：

整体流程与 NCBI 类似。


```python
[ ArrayExpress (input) ]
       ↓
[ ENA/SRA (raw data: FASTQ/BAM) ]
       ↓
[ Single Cell Expression Atlas (processed data and metadata) ]
```

* ArrayExpress：提交入口。

* ENA（European Nucleotide Archive）/SRA：存储原始测序数据。

* SCEA：接收处理后数据、metadata 和 pipelines，并将其整合进门户网站。


### 所需数据

原始数据：FASTQ 或 BAM 文件，提交至 ENA/SRA。

处理后数据：表达矩阵（genes × cells）、细胞 metadata（clusters、cell types、QC、mitochondrial percentage）、归一化文件。

脚本/pipelines：用于分析的 `.R`、`.ipynb`、`.sh` 文件。

Metadata：完整描述样本、细胞和实验条件的表格，类似 BioSample。

接受格式：`.tsv`、`.h5ad`、`.loom`，以及 `.R`、`.ipynb` 等补充文件。

### 创建 MAGE-TAB 文件

SCEA 使用 MAGE-TAB 标准记录 metadata。这是必需的，主要包含两个文件：

* IDF（Investigation Description File）：

    - 描述研究：title、abstract、contacts、相关发表文章。

    - 示例字段：Investigation Title、Experiment Description、Submitter Email。

* SDRF（Sample and Data Relationship File）：

    - 用详细表格关联样本、细胞、文件以及生物学/技术特征。

    - 重要字段：Sample Name、Organism、Cell Type、Library Prep、Sequencing Protocol、File Name（FASTQ、counting matrix 等）。

SDRF 简化示例：

|Sample Name|Organism|Tissue|Cell Type|Library Prep|Sequencing Protocol|File Name|
|------------|------------------|------------------|------------------|------------------|------------------|------------------|
|CHIKV_sc01|Homo sapiens|PBMC|lymphocyte|10x Genomics Chromium 3’ v3|Illumina NovaSeq 6000|CHIKV_sc01_R1.fastq.gz;CHIKV_sc01_R2.fastq.gz|

还应包含 dissociation 信息、capture platform（例如 10x Genomics、Smart-seq2）和实验条件。


## 提交至 ArrayExpress

ArrayExpress 是 EMBL-EBI 的转录组数据仓库，也是转录组数据提交入口。所有单细胞 RNA-seq 或空间转录组数据在整合进 SCEA 之前，都需要通过 ArrayExpress 提交。


```python
[ ArrayExpress ]
   ├── ENA (raw data: FASTQ/BAM)
   └── Expression Atlas / Single Cell Expression Atlas (processed data + metadata)
```

### 所需数据

原始数据：FASTQ 或 BAM，发送至 ENA。

处理后数据：表达矩阵（genes × cells）、细胞 metadata（clusters、cell types、QC）、归一化文件。

脚本/pipelines：`.R`、`.ipynb`、`.sh`，用于保证可重复性。

Metadata：完整描述样本、细胞和实验条件的表格。

### 创建 MAGE-TAB 文件

与前文描述相同。

### 流程

1. 在 EMBL-EBI 创建账号。

2. 准备 MAGE-TAB 文件（IDF + SDRF）。

3. 将原始数据发送至 ENA。

4. 关联 BioProject 和 BioSample IDs。

5. 向 ArrayExpress 提交 metadata 和处理后数据。

6. 上传 `.tsv`、`.h5ad`、`.loom` 和 `.rds` 文件。

7. 上传脚本/pipelines。

8. Curation：Atlas 团队审核 metadata，应用本体，并重新处理数据。

9. Publication：数据集获得公共标识符，例如 `E-MTAB-12345`，并整合进 Single Cell Expression Atlas。


### Curation 与整合

Atlas 团队会使用标准化流程重新处理数据，例如比对、归一化和聚类。

Metadata 会用本体进行统一，例如 Cell Ontology、Uberon、Disease Ontology。

数据集会获得公共标识符，例如 `E-MTAB-12345`，并可在门户网站中检索。

## **提交至 Human Cell Atlas Data Portal（HCA Data Portal）的流程**

HCA Data Portal 并不是一个“任何人都可以提交任何数据”的开放仓库；它有明确的数据接收标准。该门户接受单细胞和空间转录组数据，包括 scRNA-seq、ATAC-seq、多组学、spatial RNA-seq，也包括原始数据（FASTQ/BAM）和处理后数据（AnnData `.h5ad` 矩阵）。

数据必须附带结构化 metadata，并遵循官方 schema（Tier 1 和 Tier 2）。

此外，提交数据应来自高质量研究，具备清晰实验方案和充分文档，以便复用。

限制：

* 不接受未经整理的数据：数据集会经过技术审核，以确保一致性和质量。

* 隐私：人类数据必须匿名化。敏感信息（Tier 2，例如年龄、性别、临床状态）受控访问，只能在安全环境中使用。

* 格式：只接受标准化格式，例如 FASTQ、BAM、AnnData `.h5ad`，以及结构化表格中的 metadata。

* 范围：重点是单细胞和空间数据。例如 bulk RNA-seq 数据不纳入 HCA Data Portal。


### Tiers

Metadata 被组织为两个层级（tiers），用于区分技术信息和更敏感的信息。

#### Tier 1 Metadata：解释数据所必需的技术和实验信息

这些信息保证数据符合 FAIR，即 findable、accessible、interoperable 和 reusable。Tier 1 数据会通过 HCA Data Portal 以及 CellxGene Discover 等平台公开提供。

* 字段示例：

* Organism

* Tissue / Organ

* Cell type（使用 Cell Ontology、Uberon 等本体）

* Library preparation method（10x Genomics、Smart-seq2 等）

* Sequencing protocol（Illumina NovaSeq 等）

* File names（FASTQ、BAM、AnnData `.h5ad`）

#### Tier 2 Metadata：可能包含敏感或临床数据的补充信息

这些信息可以丰富生物学解释，同时需要维持匿名化和隐私保护。Tier 2 metadata 采用受控访问，其中部分信息可能被限制或匿名化，以保护捐赠者。

* 字段示例：

* Donor age

* Sex

* Ethnicity

* Clinical condition / disease status

* Treatment history


完整文档可在 HCA Data Portal 和数据提交指南中查看：

### 结构

HCA 将数据组织为不同层级，以保证可访问性，同时保护敏感信息：


```python
[ HCA Data Coordination Platform ]
       ├── ENA/SRA (Raw data: FASTQ)
       ├── HCA Data Repository (Tier 2 metadata + sensitive data)
       └── CellxGene Discover (matrices AnnData + Tier 1 metadata)
```

* HCA Data Repository：存储原始文件（FASTQ）和 Tier 2 metadata，其中可能包含个人或敏感信息。

* CellxGene Discover：存储 AnnData 格式（`.h5ad`）的表达矩阵和 Tier 1 metadata，例如捕获方案、细胞富集、QC 等技术信息。

* Data Coordination Platform（DCP）：提交入口。你可以在这里注册 project、提交数据集，并获得 accession ID。


## 所需数据

原始数据：FASTQ 或 BAM 文件，提交至 HCA Data Repository。

处理后数据：AnnData 格式（`.h5ad`）的表达矩阵（genes × cells）。

Tier 1 Metadata：技术信息

* Capture platform
* Dissociation method
* Library protocol
* QC

Tier 2 Metadata：更详细的信息

* Clinical data
* Age
* Sex
* Clinical condition

### 提交流程

1. 在 [HCA Data Coordination Platform](https://contribute.data.humancellatlas.org/?form=MG0AV3&utm_source=copilot.com) 注册 project。
    - 创建新 project，并提供 title、summary 和 contact information。
    - 关联已经提交至 ENA/SRA 的原始数据（FASTQ/BAM）。

2. 根据官方 ingestion guide 准备 Tier 1 和 Tier 2 metadata。

3. 将原始数据提交至 HCA Data Repository。

4. 将处理后的矩阵（AnnData `.h5ad`）提交至 HCA portal。
    - 文件会根据官方 schema 进行验证，检查必填字段等内容。
    - Curation 完成后，数据集会整合进 [CellxGene Discover](https://cellxgene.cziscience.com/discover?form=MG0AV3&utm_source=copilot.com)。

5. 获得 accession ID，并跟踪 curation 进度。

6. Publication：数据整合进门户网站，数据集获得公共标识符，例如 `HCA12345`。
    - 数据将可在 CellxGene Discover 中搜索和查看，并可按 organism、tissue、cell type、disease 等条件过滤。
